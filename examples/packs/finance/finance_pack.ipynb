A worked example, not a blank schema: the dimension schema below is the
pack's actual `schema.yaml`, and the showcase query demonstrates the
legitimate hybrid mechanism — anchors bridging a vocabulary gap that

```python
from examples.packs import load_pack, run_showcase
from kse_memory.core.projection import OnnxEmbedder

pack = load_pack("finance")
print(pack.schema_path.read_text())
```

```python
outcome = await run_showcase(pack, OnnxEmbedder())
print(f'query        : {outcome.query}')
print(f'dense rank of target : {outcome.dense_rank}  (top: {outcome.dense_top})')
print(f'hybrid rank of target: {outcome.hybrid_rank}  (top: {outcome.hybrid_top})')
assert outcome.hybrid_rank < outcome.dense_rank, 'the showcase claim'
print()
print(pack.showcase.why)
```
